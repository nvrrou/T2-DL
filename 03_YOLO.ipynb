## YOLO- Entrenamiento
___

Tuvimos que cambiar el formato de los valores del **.yaml** debido a que estaba en formato de windows y lo pasamos a Linux.

```python
from ultralytics import YOLO

# 1. Cargar el modelo base preentrenado
model = YOLO("yolov8n.pt") 

# 2. Iniciar el entrenamiento apuntando a tu archivo .yaml en K:
results = model.train(
    data="data/FjordFish/data.yaml",   # O la ruta donde tengas guardado este .yaml
    epochs=10,                 # Ajusta las épocas requeridas
    imgsz=640,
    batch=16,
    name="entrenamiento_fjordfish",
    project="runs/detect"
)

print(results)
```
